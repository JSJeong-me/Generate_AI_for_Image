## Sampler 지정하여 이미지 생성하기

```python
# 설치 (필요시)
# !pip install diffusers transformers accelerate --quiet
```

```python
from diffusers import StableDiffusionPipeline, DDIMScheduler, EulerDiscreteScheduler
import torch
```

```python
# 모델 로드 (Stable Diffusion v1.5)
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
)
pipe = pipe.to("cuda")
```

```python
# 1. DDIM Sampler 사용
pipe.scheduler = DDIMScheduler.from_config(pipe.scheduler.config)
image_ddim = pipe("A cat sitting on a chair").images[0]
image_ddim.save("cat_ddim.png")
```

```python
# 2. Euler Sampler 사용
pipe.scheduler = EulerDiscreteScheduler.from_config(pipe.scheduler.config)
image_euler = pipe("A cat sitting on a chair").images[0]
image_euler.save("cat_euler.png")
```

## 지금부터는 CFG Scale 입력 받아 Drawing 합니다

```python
# 라이브러리 설치 (필요시)
!pip install diffusers transformers accelerate --quiet
```

```python
from diffusers import StableDiffusionPipeline
import torch
```

```python
# 1. 사용자 입력 (프롬프트와 CFG scale)
prompt = input("프롬프트를 입력하세요: ")
cfg_scale = float(input("CFG scale 값을 입력하세요 (예: 7.5): "))
```

```python
# 2. 파이프라인 로드
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16,
)
pipe = pipe.to("cuda")
```

```python
# 3. 이미지 생성
with torch.autocast("cuda"):
    image = pipe(prompt, guidance_scale=cfg_scale).images[0]
```

```python
# 4. 결과 출력 및 저장
image.show()
image.save("result.png")
```

## Step 입력 기능 추가

```python
# 설치 (필요시)
# !pip install diffusers transformers accelerate --quiet

from diffusers import StableDiffusionPipeline
import torch
```

```python

# 1. 사용자 입력
prompt = input("프롬프트를 입력하세요: ")
cfg_scale = float(input("CFG scale 값을 입력하세요 (예: 7.5): "))
steps = int(input("샘플링 step 값을 입력하세요 (예: 30): "))
```

```python

# 2. 모델 로드
# pipe = StableDiffusionPipeline.from_pretrained(
#     "runwayml/stable-diffusion-v1-5",
#     torch_dtype=torch.float16,
# )
# pipe = pipe.to("cuda")

# 3. 이미지 생성
with torch.autocast("cuda"):
    image = pipe(
        prompt,
        guidance_scale=cfg_scale,
        num_inference_steps=steps
    ).images[0]

# 4. 이미지 출력 및 저장
image.show()
image.save("result.png")
```
